Compute the standard Deviations on a data set and justify
which set has a larger spread and compute covariance.

```python
import numpy as np
import pandas as pd
dataset1 = [10, 12, 15, 18, 20]
dataset2 = [2, 4, 8, 12, 16]

std_dev1 = np.std(dataset1)
std_dev2 = np.std(dataset2)

print("Standard Deviation of Dataset 1:", std_dev1)
print("Standard Deviation of Dataset 2:", std_dev2)

covariance = np.cov(dataset1, dataset2)[0, 1]
print("Covariance:", covariance)
```

```python
import numpy as np

# Sample datasets
data_1 = [10, 12, 23, 23, 16, 23, 21, 16]
data_2 = [32, 45, 37, 48, 35, 50, 41, 39]

# Step 1: Compute Mean
mean1 = np.mean(data_1)
mean2 = np.mean(data_2)

# Step 2: Compute Standard Deviation
std_1 = np.std(data_1, ddof=1)  # Sample std deviation
std_2 = np.std(data_2, ddof=1)

# Step 3: Compute Covariance
covariance_matrix = np.cov(data_1, data_2)
covariance = covariance_matrix[0][1]

# Step 4: Output Results
print(f"Mean of Data_1: {mean1:.2f}")
print(f"Mean of Data_2: {mean2:.2f}")
print(f"Standard Deviation of Data_1: {std_1:.2f}")
print(f"Standard Deviation of Data_2: {std_2:.2f}")
print(f"Covariance between Data_1 and Data_2: {covariance:.2f}")

# Step 5: Justification of Spread
if std_1 > std_2:
    print("Data_1 has a larger spread.")
elif std_2 > std_1:
    print("Data_2 has a larger spread.")
else:
    print("Both datasets have equal spread.")
```
