```python
import numpy as np

def mcculloch_pitts(inputs, weights, threshold):
  """Implements the McCulloch-Pitts neuron model.

  Args:
    inputs: A list of input values (0 or 1).
    weights: A list of weights corresponding to each input.
    threshold: The activation threshold.

  Returns:
    The output of the neuron (0 or 1).
  """
  weighted_sum = np.dot(inputs, weights)
  if weighted_sum >= threshold:
    return 1
  else:
    return 0

# Define gate parameters
and_weights = [1, 1]
or_weights = [1, 1]
not_weights = [-1]
threshold = 1

# Test the gates
print("AND Gate:")
print(mcculloch_pitts([0, 0], and_weights, threshold))  # Output: 0
print(mcculloch_pitts([0, 1], and_weights, threshold))  # Output: 0
print(mcculloch_pitts([1, 0], and_weights, threshold))  # Output: 0
print(mcculloch_pitts([1, 1], and_weights, threshold))  # Output: 1

print("\nOR Gate:")
print(mcculloch_pitts([0, 0], or_weights, threshold))  # Output: 0
print(mcculloch_pitts([0, 1], or_weights, threshold))  # Output: 1
print(mcculloch_pitts([1, 0], or_weights, threshold))  # Output: 1
print(mcculloch_pitts([1, 1], or_weights, threshold))  # Output: 1

print("\nNOT Gate:")
print(mcculloch_pitts([0], not_weights, threshold))  # Output: 1
print(mcculloch_pitts([1], not_weights, threshold))  # Output: 0
```

```python
import numpy as np

# Step 1: Define the Threshold Activation Function
def threshold_activation(net_input, threshold):

  return np.where(net_input >= threshold, 1, 0)

# Step 2: Implement McCulloch-Pitts Neuron for AND Function
def mcp_neuron_and(inputs):

  weights = np.array([1, 1]) # Weights for inputs
  threshold = 2 # Threshold for AND function

# Compute weighted sum
  net_input = np.dot(inputs, weights)

# Apply threshold activation function
  output = threshold_activation(net_input, threshold)
  return output

# Step 3: Test AND Function with All Possible Inputs
input_data = np.array([[0, 0], [0, 1], [1, 0], [1, 1]]) # All combinations of binary inputs
outputs = np.array([mcp_neuron_and(x) for x in input_data]) # Get outputs

# Step 4: Display Input-Output Truth Table
print("McCulloch-Pitts Neuron for AND Function")
print("-" * 30)
print("Input 1 | Input 2 | Output")
print("-" * 30)
for i, (x1, x2) in enumerate(input_data):
  print(f" {x1} | {x2} | {outputs[i]}")

  print("-" * 30)
```
